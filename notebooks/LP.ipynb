Escreva uma função que aceite um inteiro não negativo *n* e uma string *s* como parâmetros e retorne uma string de vezes *s* exatas repetidas *n* vezes.

```python
def repeat_str(repeat, string):
    if not isinstance(repeat, int):
        raise TypeError("Observação: repeat deve ser inteiro.")
    if not isinstance(string, str):
        raise TypeError("Observação: string deve ser texto.")
    return repeat * string

print(repeat_str(2, "hi"))
```

Crie uma função que receba um inteiro como argumento e retorne "Even" para números pares e "Odd" para números ímpares.

```python
def even_or_odd(number):
    if type(number) is not int:
        raise TypeError("Observação: number deve ser inteiro.")
    return "Even" if number%2 == 0 else "Odd"
    
print(even_or_odd(2))
```

Muito simples, dado um número, encontre seu oposto (inverso aditivo).

```python
def opposite(number):
    return -number

print(opposite(0.2))
```

É bem direto. Seu objetivo é criar uma função que remova o primeiro e o último caractere de uma string. Você recebe um parâmetro, a string original. Você não precisa se preocupar com strings com menos de dois caracteres.

```python
def remove(string):
    return string[1:-1]

print(remove("Lais"))
```
