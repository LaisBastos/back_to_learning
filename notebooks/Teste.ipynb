```python
df = pd.DataFrame({
    "Dia": [1, 2, 3, 4, 5],
    "Paradas": [3, 1, 4, 2, 1]
})

plt.plot(df["Dia"], df["Paradas"], marker="o")
plt.title("Paradas por dia")
plt.xlabel("Dia")
plt.ylabel("Quantidade")
plt.grid(True)
plt.show()
```
